```python
from pathlib import Path

p = Path("src/llmbench/backends/vllm_backend.py")
s = p.read_text(encoding="utf-8")

old1 = """        self.engine: Any = None
        self.SamplingParams: Any = None
"""
new1 = """        self.engine: Any = None
        self.SamplingParams: Any = None
        self.loop: Any = None
"""
assert old1 in s
s = s.replace(old1, new1, 1)

old2 = """        self.engine = AsyncLLMEngine.from_engine_args(args)
        self.loaded = True
"""
new2 = """        self.loop = asyncio.new_event_loop()
        asyncio.set_event_loop(self.loop)

        async def build_engine() -> Any:
            return AsyncLLMEngine.from_engine_args(args)

        try:
            self.engine = self.loop.run_until_complete(build_engine())
        except Exception:
            self.loop.close()
            self.loop = None
            raise
        self.loaded = True
"""
assert old2 in s
s = s.replace(old2, new2, 1)

old3 = """    @staticmethod
    def _run(coro: Any) -> Any:
        return asyncio.run(coro)
"""
new3 = """    def _run(self, coro: Any) -> Any:
        if self.loop is None:
            return asyncio.run(coro)
        return self.loop.run_until_complete(coro)
"""
assert old3 in s
s = s.replace(old3, new3, 1)

old4 = """    def close(self) -> None:
        try:
            if self.engine is not None:
                self.engine.shutdown()
        except Exception:
            pass
        super().close()
"""
new4 = """    def close(self) -> None:
        try:
            if self.engine is not None:
                self.engine.shutdown()
        except Exception:
            pass
        try:
            if self.loop is not None:
                self.loop.close()
        except Exception:
            pass
        self.loop = None
        self.engine = None
        super().close()
"""
assert old4 in s
s = s.replace(old4, new4, 1)

p.write_text(s, encoding="utf-8")
print("vLLM loop patch applied")

```

```python
from llmbench.config import load_config
from llmbench.backends import build_backend
from llmbench.backends.base import GenerationRequest
from llmbench.data import make_workload_prompts

cfg = load_config("configs/benchmark.yaml")
b = build_backend("vllm", cfg.backend("vllm"), cfg)
b.load()
p = make_workload_prompts(1, cfg.benchmark.input_tokens, cfg.benchmark.seed)[0]
r = b.generate([GenerationRequest(prompt=p, request_id=1, max_new_tokens=16, temperature=0.0, top_p=1.0)])[0]
print("ERROR:", r.error)
print("TOKENS:", r.output_tokens, "TEXT:", repr(r.text[:200]))
b.close()
```

```python
import json, glob, os

ok = True
for p in sorted(glob.glob("results_smoke/raw/benchmark_*.json")):
    rows = json.load(open(p, encoding="utf-8"))
    e = sum(r.get("errors", 0) for r in rows)
    t = sum(r.get("output_tokens", 0) for r in rows)
    g = all(r.get("peak_gpu_mib") is not None for r in rows)
    print(os.path.basename(p), "errors=", e, "out_tokens=", t, "gpu_ok=", g)
    ok &= (e == 0 and t > 0 and g)

print("SMOKE_OK =", ok)
```
